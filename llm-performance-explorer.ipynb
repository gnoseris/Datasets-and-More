# AI Model Performance Explorer
## From Training Metrics to Arena Benchmarks

This notebook demonstrates how to use the AI Model Training Metrics + Arena Performance Benchmarks dataset.

**Dataset Sources:**
- Epoch AI: Training metrics (parameters, compute, hardware, costs)
- LMSYS Chatbot Arena: Performance benchmarks (ELO ratings, MT-Bench, MMLU)

**What You'll Learn:**
- Data exploration techniques for AI model datasets
- Performance correlation analysis
- Organization and efficiency comparisons
- Creating insightful visualizations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Configuration
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.3f' % x)

print("Libraries imported successfully!")
```

```python
# Load dataset
DATASET_PATH = "/kaggle/input/datasets/bernardinozaganellii/llm-training-metrics-arena-performance-data/"
df = pd.read_csv(DATASET_PATH + "Model Training Metrics Arena Performance Benchmarks.csv")

# Basic information
print(f"Dataset Shape: {df.shape}")
print(f"Number of Features: {len(df.columns)}")
print(f"\nData Source Distribution:")
print(df['data_source'].value_counts())
print(f"\nFirst 3 rows:")
df.head(3)
```

## Dataset Overview

Key columns include:
- **Model**: Model name
- **Organization**: Developing organization
- **Parameters**: Number of parameters
- **Training compute (FLOP)**: Training compute
- **mt_bench_score**: MT-Bench evaluation score
- **arena_elo_rating**: Arena ELO rating
- **mmlu_score**: MMLU benchmark score
- **data_source**: Origin (epoch_only, arena_only, both)

```python
# Check for missing values in key columns
key_columns = ['Model', 'Organization', 'Parameters', 'mt_bench_score', 
               'arena_elo_rating', 'mmlu_score', 'Training compute (FLOP)']

missing_info = df[key_columns].isnull().sum()
missing_pct = (missing_info / len(df) * 100).round(1)

quality_df = pd.DataFrame({
    'Missing Count': missing_info,
    'Missing %': missing_pct,
    'Available': len(df) - missing_info
})

print("Data Quality Summary:")
quality_df
```

## Top Organizations by Model Count

Which organizations are most represented in the dataset?

```python
# Top 15 organizations
org_counts = df['Organization'].value_counts().head(15)

fig = px.bar(x=org_counts.values, y=org_counts.index, 
             orientation='h',
             title='Top Organizations by Number of Models',
             labels={'x': 'Number of Models', 'y': 'Organization'},
             color=org_counts.index,
             color_discrete_sequence=px.colors.sequential.Blues_r)

fig.update_layout(height=600, showlegend=False)
fig.show()
```

## Model Size Distribution

Distribution of model parameters across the dataset.

```python
# Filter out null parameters
params_data = df['Parameters'].copy()
params_data = params_data[params_data.notna()]
params_data = params_data[params_data > 0]  # Remove any zero/negative values
params_data = df['Parameters'].dropna()

# Create histogram with explicit binning
fig = px.histogram(x=params_data, log_x=True, nbins=50,
                   title='Distribution of Model Parameters (Log Scale)',
                   labels={'x': 'Parameters (log scale)', 'y': 'Number of Models'},
                   color_discrete_sequence=['#636EFA'])

# Ensure axes are properly displayed
fig.update_xaxes(type='log')
fig.update_layout(
    showlegend=False, 
    height=500
)
fig.show()

# Summary statistics
print("Parameter Statistics:")
print(f"Filtered data length: {len(params_data)}")
print(f"Min value: {params_data.min()}")
print(f"Max value: {params_data.max()}")

# Add range specification


# Run this to check your data
print("Parameters column analysis:")
print(f"Column exists: {'Parameters' in df.columns}")
print(f"Data type: {df['Parameters'].dtype}")
print(f"Null count: {df['Parameters'].isna().sum()}")
print(f"Non-null count: {df['Parameters'].notna().sum()}")

# Check actual values
print("\nFirst 10 raw values:")
for i, val in enumerate(df['Parameters'].head(10)):
    print(f"  {i}: {repr(val)} (type: {type(val).__name__})")

# Check if conversion works
params_numeric = pd.to_numeric(df['Parameters'], errors='coerce')
print(f"\nAfter numeric conversion:")
print(f"Valid numbers: {params_numeric.notna().sum()}")
print(f"NaN from conversion: {params_numeric.isna().sum()}")

# Check value ranges
valid_params = params_numeric.dropna()
if len(valid_params) > 0:
    print(f"\nValue range:")
    print(f"Min: {valid_params.min()}")
    print(f"Max: {valid_params.max()}")
    print(f"Mean: {valid_params.mean()}")
else:
    print("\nNo valid numeric parameters found!")
```

## Performance Benchmark Correlations

How do different benchmark scores relate to each other?

```python
# Select performance columns
perf_cols = ['mt_bench_score', 'arena_elo_rating', 'mmlu_score']
perf_data = df[perf_cols].dropna()

# Calculate correlations
corr_matrix = perf_data.corr()

# Create correlation heatmap
fig = go.Figure(data=go.Heatmap(
    z=corr_matrix.values,
    x=['MT-Bench', 'Arena ELO', 'MMLU'],
    y=['MT-Bench', 'Arena ELO', 'MMLU'],
    colorscale='RdBu',
    zmin=-1, zmax=1,
    text=corr_matrix.values.round(3),
    texttemplate="%{text}"
))

fig.update_layout(
    title='Correlation Between Performance Benchmarks',
    xaxis_title='Benchmark',
    yaxis_title='Benchmark',
    width=500, height=400
)
fig.show()
```

```python
# Scatter plot: Parameters vs MT-Bench Score
scatter_data = df[df['Parameters'].notna() & df['mt_bench_score'].notna()]

fig = px.scatter(scatter_data, 
                 x='Parameters', 
                 y='mt_bench_score',
                 color='Organization',
                 hover_data=['Model'],
                 log_x=True,
                 title='Model Parameters vs MT-Bench Score',
                 labels={'Parameters': 'Parameters (log scale)', 
                        'mt_bench_score': 'MT-Bench Score'},
                 opacity=0.7)

fig.update_traces(marker=dict(size=8))
fig.update_layout(height=600)
fig.show()
```

## Model Efficiency Analysis

Which models deliver the best performance relative to their size?

```python
# Calculate efficiency (score per billion parameters)
efficiency_data = df[df['Parameters'].notna() & df['mt_bench_score'].notna()].copy()
efficiency_data['efficiency'] = efficiency_data['mt_bench_score'] / (efficiency_data['Parameters'] / 1e9)

# Top 10 most efficient models
top_efficient = efficiency_data.nlargest(10, 'efficiency')[['Model', 'Organization', 
                                                           'Parameters', 'mt_bench_score', 'efficiency']]

print("Top 10 Most Efficient Models (MT-Bench Score per Billion Parameters):")
top_efficient.round(3)
```

```python
fig = px.bar(top_efficient, 
             x='efficiency', 
             y='Model',
             color='Organization',
             orientation='h',
             title='Top 10 Most Efficient Models',
             labels={'efficiency': 'Efficiency (MT-Bench Score / Billion Params)'},
             color_continuous_scale='Viridis')

fig.update_layout(height=500, yaxis={'categoryorder':'total ascending'})
fig.show()
```

```python
# Compare Arena ELO vs MT-Bench
comparison_data = df[df['arena_elo_rating'].notna() & df['mt_bench_score'].notna()]

fig = px.scatter(comparison_data,
                 x='arena_elo_rating',
                 y='mt_bench_score',
                 color='Organization',
                 hover_data=['Model'],
                 title='Arena ELO Rating vs MT-Bench Score',
                 labels={'arena_elo_rating': 'Arena ELO Rating',
                        'mt_bench_score': 'MT-Bench Score'})

fig.update_traces(marker=dict(size=10, opacity=0.7))
fig.update_layout(height=600)
fig.show()
```
